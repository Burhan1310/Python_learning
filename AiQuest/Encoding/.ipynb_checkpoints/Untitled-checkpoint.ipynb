# Sklearn Preprocessing

# Label Encoder

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df1.Area = le.fit_transform(df[['Area']])
```

```python
df1.head()
```

```python
le.classes_
```

```python
le.transform(le.classes_)
```

**Decode the encoded values back to original classes**

```python
decoded_classes = le.inverse_transform(df1.Area)
decoded_classes
```

```python
df1.head()
```

```python
df1.Area = decoded_classes
```

```python
df1.head()
```

# Mean Encoding

```python
df6.head()
```

```python
mean_encoded_area = df6.groupby('Area')['Profit'].mean().to_dict()
mean_encoded_area
```

```python
unique_areas = df6['Area'].unique()
unique_areas
```

```python
df6.Area = df6.Area.map(mean_encoded_area)
df6.head()
```

# Frequency Encoding

```python
df7.head()
```

```python
frequency_encoding = df7['Area'].value_counts().to_dict()
frequency_encoding
```

```python
df7['Area_frequency_Encoded'] = df.Area.map(frequency_encoding)
df7.head()
```
