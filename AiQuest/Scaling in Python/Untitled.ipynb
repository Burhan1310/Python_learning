```python
lst = [100, 300, 250, 450, 150, 250, 500]
```

```python
lst
```

```python
import pandas as pd
df = pd.DataFrame(lst, columns=['cost'])
df
```

# Techniques to perform Feature Transformation:

1. Normalization <br>
2. Standardization <br>
3. Log Transformation <br>
4. Robust Scaler <br>
5. Max Absolute Scaler

# Normalization / min max scaler

```python
min_cost = df.cost.min()
max_cost = df.cost.max()
```

```python
min_cost
```

```python
max_cost
```

```python
df['cost_norm_manual'] = (df.cost - min_cost) / (max_cost - min_cost)
```

```python
df
```

<b>Using Sklearn Library

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
mms = MinMaxScaler()  # by default feature_range=(0, 1)
```

```python
df['cost_norm_sklearn'] = mms.fit_transform(df[['cost']])
```

```python
df
```

# Standardization / Standard Scaler

```python
mean_cost = df.cost.mean()
std = df.cost.std()
```

```python
std
```

```python
df['cost_std_manual'] = (df.cost - mean_cost) / std
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
df['cost_std_sklearn'] = scaler.fit_transform(df[['cost']])
```

```python
df
```

```python
std_ddof = df.cost.std(ddof=0)  ### Delta Degree of freedom (n-1)
```

```python
std_ddof
```

```python
std_ddof1 = df.cost.std(ddof = 1)
```

```python
std_ddof1
```

```python
df['cost_std_ddof0'] = (df.cost - mean_cost) / std_ddof
```

```python
df
```

```python
df['cost_std_ddof1'] = (df.cost - mean_cost) / std_ddof1
```

```python
df
```

# Log transformer

```python
from sklearn.preprocessing import FunctionTransformer
import numpy as np
```

```python
ft = FunctionTransformer(np.log1p)
df['cost_ft'] = ft.fit_transform(df[['cost']])
df
```

# Robust Scaler

```python
from sklearn.preprocessing import RobustScaler
```

```python
rs = RobustScaler()
df['cost_robust_scaler'] = rs.fit_transform(df[['cost']])
df
```

# Max Abs Scaler

```python
from sklearn.preprocessing import MaxAbsScaler
```

```python
mas = MaxAbsScaler()
```

```python
df['cost_max_abs'] = mas.fit_transform(df[['cost']])
```

```python
df
```
