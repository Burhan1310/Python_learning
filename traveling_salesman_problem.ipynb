```python
from sys import maxsize
```

```python
v = 4
```

```python
def traveling_salesman_function(graph, starting_point):
  vertex = []
  for i in range(v):
    if i != starting_point:
      vertex.append(i)

  min_path = maxsize
  while True:
    current_cost = 0
    k = starting_point
    for i in range(len(vertex)):
      current_cost += graph[k][vertex[i]]
      k = vertex[i]
    current_cost += graph[k][starting_point]
    min_path = min(min_path, current_cost)

    if not next_permutation(vertex):
      break

  return min_path
```

```python
def next_permutation(lst):
  n = len(lst)
  i = n-2

  while i >= 0 and lst[i] > lst[i+1]:
    i -= 1

  if i == -1:
    return False

  j = i+1
  while j < n and lst[j] > lst[i]:
    j += 1

  j -= 1
  lst[i], lst[j] = lst[j], lst[i]
  left = i+1
  right = n-1

  while left < right:
    lst[left], lst[right] = lst[right], lst[left]
    left += 1
    right -= 1
  return True
```

```python
graph = [[0,10,15,20],[10,0,35,25],[15,35,0,30],[20,25,30,0]]
starting_point = 0
```

```python
traveling_salesman_function(graph, starting_point)
```

### Explanation of the `next_permutation` Function



The `next_permutation` function generates the next lexicographical permutation of a list. This is useful for systematically trying all possible orders of elements (like cities in TSP).



**How it works:**

1. **Find the rightmost pair where the left element is less than the right:**

   - Start from the second last element (`i = n-2`) and move left until you find `lst[i] < lst[i+1]`.

   - If no such pair is found, the list is in descending order and there are no more permutations, so return `False`.



2. **Find the smallest element to the right of `i` that is greater than `lst[i]`:**

   - Start from `i+1` and move right to find such an element (`lst[j] > lst[i]`).



3. **Swap `lst[i]` and `lst[j]`:**

   - This step increases the permutation just enough to get the next one in order.



4. **Reverse the sublist after position `i`:**

   - This ensures the next permutation is the smallest possible after the swap.



5. **Return `True`:**

   - Indicates that the next permutation was successfully generated.



This algorithm is efficient and commonly used to generate all permutations in order.
