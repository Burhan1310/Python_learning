```python
import torch
```

```python
x = torch.randn(3, requires_grad=True)
x
```

```python
y = x + 2
y
```

```python
z = y * y * 2
z
```

```python
z = z.mean()
```

```python
z.backward()
```

```python
x.grad
```

```python
weights = torch.ones(4, requires_grad=True)
weights
```

```python
for epoch in range(3):
  model_output = (weights*3).sum()
  model_output.backward()

  print(weights.grad)
```

```python
for epoch in range(3):
  model_output = (weights*3).sum()
  model_output.backward()

  print(weights.grad)
  weights.grad.zero_()
```

```python
optimizer = torch.optim.SGD(weights, lr=0.01)
optimizer.step()
optimizer.zero_grad()
```
