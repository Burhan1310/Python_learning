```python
import numpy as np
```

```python
list = [3, 4, 5, 6]
arr = np.array(list)
```

```python
arr
```

```python
arr.dtype
```

**Convert Numpy to Pytorch tensors**

```python
tensors = torch.from_numpy(arr)
tensors
```

```python
# Indexing similar to numpy
tensors[:2]
```

```python
## Disadvantages of from_numpy. The array and tensor uses the same memory location
tensors[3] = 100
```

```python
tensors
```

```python
arr
```

```python
### Prevent this by using torch.tensor
tensor_arr = torch.tensor(arr)
tensor_arr
```

```python
tensor_arr[3] = 120
print(tensor_arr)
print(arr)
```

**Arithmetic** **Operation**

```python
a = torch.tensor([3,4,5], dtype=torch.float)
b = torch.tensor([4,5,6], dtype=torch.float)
print(a + b)
```

```python
torch.add(a,b)
```

```python
c = torch.zeros(3)
```

```python
c
```

```python
torch.add(a,b,out=c)
```

```python
c
```

```python
torch.add(a,b).sum()
```

**Dot Products and Multiplication Opearation**

```python
x = torch.tensor([3,4,5],dtype=torch.float)
y = torch.tensor([4,5,6], dtype=torch.float)
```

```python
x.mul(y)
```

```python
x.dot(y)
```

**Matrix Multiplication**

```python
x = torch.tensor([[1,4,2],[1,5,5]], dtype=torch.float)
y = torch.tensor([[5,7],[8,6],[9,11]],dtype=torch.float)
```

```python
torch.matmul(x,y)
```

```python
torch.mm(x,y)
```

```python
x@y
```
